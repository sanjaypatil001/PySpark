QN.You have a dataset containing employee details with their skills. The dataset includes the following columns: EmpId (Employee ID), EmpName (Employee Name), and Skill (Employee's Skill). Write a PySpark code to concatenate the skills of each employee into a comma-separated string, grouped by EmpId and EmpName. The output should list each employee along with their comma-separated list of skills.

```
from pyspark.sql import SparkSession
from pyspark.sql.functions import collect_list, concat_ws

# Initialize SparkSession
spark = SparkSession.builder \
    .appName("CommaSeparatedSkills") \
    .getOrCreate()

# Sample data
data = [
    (1, 'Joey', 'ADF'), (1, 'Joey', 'ADB'), (1, 'Joey', 'Python'),
    (2, 'Rachel', 'ADF'), (2, 'Rachel', 'SQL'), (2, 'Rachel', 'JAVA'),
    (3, 'Victor', 'ADF'), (3, 'Victor', 'SQL'), (3, 'Victor', 'DB'),
    (4, 'Mike', 'SQL'), (4, 'Mike', 'DB'), (4, 'Mike', 'SSAS'), (4, 'Mike', 'ADF')
]

# Define schema
schema = ["EmpId", "EmpName", "Skill"]

# Create DataFrame
df = spark.createDataFrame(data, schema)

# Group by EmpId and EmpName, and concatenate the skills into a comma-separated list
result_df = df.groupBy("EmpId", "EmpName") \
    .agg(concat_ws(",", collect_list("Skill")).alias("Skills"))

# Show the result
result_df.show(truncate=False)
```
