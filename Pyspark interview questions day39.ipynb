```
#How to fill null values in pyspark

from pyspark.sql import SparkSession

# Initialize Spark session
spark = SparkSession.builder.appName("FillNullValues").getOrCreate()

# Sample data with null values
data = [
    (1, "Amy", None),
    (2, None, 25),
    (3, "Jack", None),
    (4, None, 30),
]
columns = ["id", "name", "age"]

# Create DataFrame
df = spark.createDataFrame(data, columns)
df.show()

# Fill null values
df_filled = df.fillna("none")  # Fills all string columns with 'none'
df_filled = df_filled.fillna(0)   # Fills all numeric columns with 0

# Show the result
df_filled.show()
```
