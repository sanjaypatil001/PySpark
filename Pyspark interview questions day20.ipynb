```
#Write a PySpark program with sample data and schema to calculate the number of visitors to a website.

from pyspark.sql import SparkSession
from pyspark.sql.types import StructType, StructField, StringType

# Create a SparkSession
spark = SparkSession.builder.appName("WebsiteVisitors").getOrCreate()

# Define schema for the website visit data
schema = StructType([
    StructField("Name", StringType(), True),
    StructField("VisitedWebsite", StringType(), True)
])
# Sample log data representing website visits
data = [
    ("Arwen", "Yes"),
    ("Bilbo", "Yes"),
    ("Nick", "No"),
    ("Frodo", "Yes"),
    ("Merry", "No"),
    ("Luca", "Yes")
]

# Create a DataFrame using the data and schema
visitors_df = spark.createDataFrame(data, schema=schema)

# Show the contents of the visitors DataFrame
visitors_df.show()

# Calculate the total number of visitors who visited the website
visitors_count = visitors_df.filter(visitors_df.VisitedWebsite == "Yes").count()

print(f"Number of visitors to the website: {visitors_count}")
```
