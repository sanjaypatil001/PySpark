```
#Qn.Write a pyspark code to find the topmost department having highest salary

from pyspark.sql import SparkSession
from pyspark.sql.functions import col, max

# Initialize SparkSession
spark = SparkSession.builder \
    .appName("TopDepartmentSalary") \
    .getOrCreate()

# Sample data
data = [
    (1, "HR", "John", 5000),
    (2, "Engineering", "Monica", 8000),
    (1, "HR", "Max", 10000),
    (3, "Finance", "Amy", 7000),
    (2, "Engineering", "Lego", 12000),
    (3, "Finance", "Tim", 9000)
]

# Define schema and create DataFrame
columns = ["department_id", "department_name", "employee_name", "salary"]
df = spark.createDataFrame(data, schema=columns)

# Find the department with the highest salary
# First, find the highest salary
highest_salary = df.select(max("salary")).collect()[0][0]

# Filter departments with the highest salary
top_department_df = df.filter(col("salary") == highest_salary) \
                       .select("department_name", "salary") \
                       .distinct()

# Show the department with the highest salary
top_department_df.show()
```
