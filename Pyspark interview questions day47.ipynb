```
#Qn.Write a Pyspark code to calculate the avg processing time of the machine

from pyspark.sql import SparkSession
from pyspark.sql.functions import col

# Initialize SparkSession
spark = SparkSession.builder \
    .appName("AverageProcessingTime") \
    .getOrCreate()

# Sample data
data = [
    (0, 0, 'start', 0.712), (0, 0, 'end', 1.520),
    (0, 1, 'start', 3.140), (0, 1, 'end', 4.120),
    (1, 0, 'start', 0.550), (1, 0, 'end', 1.550),
    (1, 1, 'start', 0.430), (1, 1, 'end', 1.420),
    (2, 0, 'start', 4.100), (2, 0, 'end', 4.512),
    (2, 1, 'start', 2.500), (2, 1, 'end', 5.000)
]

# Define schema
schema = ["Machine_id", "processid", "activityid", "timestamp"]

# Create DataFrame
df = spark.createDataFrame(data, schema)

# Filter out the 'start' and 'end' activities
start_df = df.filter(col("activityid") == "start").withColumnRenamed("timestamp", "start_timestamp")
end_df = df.filter(col("activityid") == "end").withColumnRenamed("timestamp", "end_timestamp")

# Join start and end based on Machine_id and processid
joined_df = start_df.join(end_df, on=["Machine_id", "processid"], how="inner")

# Calculate processing time for each process
processing_time_df = joined_df.withColumn("processing_time", col("end_timestamp") - col("start_timestamp"))

# Calculate average processing time for each machine
average_processing_time_df = processing_time_df.groupBy("Machine_id").avg("processing_time")

# Show the result
average_processing_time_df.show()
```
