```
#Calculate the total revenue generated from all orders
#1: Total revenue generated from all orders
#2: Top 5 orders with highest total amount

from pyspark.sql import SparkSession
from pyspark.sql.functions import col

# Create a Spark session
spark = SparkSession.builder.appName("OrderProductAnalysis").getOrCreate()

# Load the datasets
order_data = [
    (1, "C101", "2023-07-01", 150),
    (2, "C102", "2023-07-02", 200),
    (3, "C101", "2023-07-02", 100),
    (4, "C103", "2023-07-03", 300),
    (5, "C102", "2023-07-04", 250),
    (6, "C101", "2023-07-05", 120)
]
order_columns = ["order_id", "customer_id", "order_date", "total_amount"]
orders_df = spark.createDataFrame(order_data, order_columns)

product_data = [
    (1, "Product A", 500),
    (2, "Product B", 50),
    (3, "Product C", 800),
    (4, "Product D", 30),
    (5, "Product E", 75)
]
product_columns = ["product_id", "product_name", "price"]
products_df = spark.createDataFrame(product_data, product_columns)

# Question 1: Total revenue generated from all orders
total_revenue = orders_df.selectExpr("sum(total_amount) as total_revenue").collect()[0]["total_revenue"]
print("Total Revenue:", total_revenue)

# Question 2: Top 5 orders with highest total amount
top_orders = orders_df.orderBy(col("total_amount").desc()).limit(5)
top_orders.show()
```
