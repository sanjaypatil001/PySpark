```
#Find the person's start and end destination of the round trip.

from pyspark.sql import SparkSession
from pyspark.sql.functions import col, first, last

# Initialize Spark session
spark = SparkSession.builder.appName("RoundTrip").getOrCreate()

# Sample data
data = [
    ("Joey", 1, "BANGALORE", "HYDERABAD"),
    ("Joey", 2, "HYDERABAD", "CHENNAI"),
    ("Joey", 3, "CHENNAI", "BANGALORE"),
    ("Ross", 1, "MUMBAI", "PUNE"),
    ("Ross", 2, "PUNE", "MUMBAI"),
]

# Create DataFrame
columns = ["name", "trip_id", "source", "destination"]
df = spark.createDataFrame(data, columns)

# Find start and end destinations using window functions
from pyspark.sql.window import Window

# Define a window specification for each person ordered by trip_id
window_spec = Window.partitionBy("name").orderBy("trip_id")

# Aggregate to get start and end destinations
result = df.groupBy("name") \
           .agg(first("source").alias("start"), last("destination").alias("end"))

# Show the result
result.show()
```
