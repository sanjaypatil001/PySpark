```
#Write a PySpark code to find the names of all activities that have neither the maximum nor the minimum number of participants?

from pyspark.sql import SparkSession
from pyspark.sql.functions import col, max as spark_max, min as spark_min

# Initialize Spark session
spark = SparkSession.builder.appName("ActivityParticipants").getOrCreate()

# Sample data
data = [
    ("Yoga", 50),
    ("Swimming", 30),
    ("Dance", 70),
    ("Cycling", 20),
    ("Running", 100),
]
columns = ["activity", "participants"]

# Create DataFrame
df = spark.createDataFrame(data, columns)

# Find the maximum and minimum number of participants
max_participants = df.agg(spark_max("participants")).collect()[0][0]
min_participants = df.agg(spark_min("participants")).collect()[0][0]

# Filter activities with neither maximum nor minimum participants
filtered_activities = df.filter(
    (col("participants") != max_participants) & (col("participants") != min_participants)
)

# Show the result
filtered_activities.show()
```
