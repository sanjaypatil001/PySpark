Qn.You have a dataset that contains employee information with the columns Department and Gender. Write a PySpark code to obtain a report that contains the following details for each department:

DEPARTMENT NAME

TOTALEMPLOYEES (the total number of employees in the department)

MALE EMPLOYEES (the number of male employees in the department)

FEMALE EMPLOYEES (the number of female employees in the department)

```
from pyspark.sql import SparkSession
from pyspark.sql.functions import col, when, count

# Initialize SparkSession
spark = SparkSession.builder \
    .appName("DepartmentEmployeeCount") \
    .getOrCreate()

# Sample dataset
data = [
    ('HR', 'Male'),
    ('HR', 'Female'),
    ('IT', 'Male'),
    ('IT', 'Female'),
    ('IT', 'Male'),
    ('Sales', 'Female'),
    ('Sales', 'Male'),
    ('Sales', 'Female')
]

# Define schema
schema = ["Department", "Gender"]

# Create DataFrame
df = spark.createDataFrame(data, schema)

# Count total employees, male employees, and female employees per department
result_df = df.groupBy("Department") \
    .agg(
        count("*").alias("TotalEmployees"),
        count(when(col("Gender") == "Male", 1)).alias("MaleEmployees"),
        count(when(col("Gender") == "Female", 1)).alias("FemaleEmployees")
    )

# Show the result
result_df.show()
```
