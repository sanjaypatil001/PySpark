```
#Write a PySpark program to process a dataset containing names and their hobbies (stored as a comma-separated string). The program should split the hobbies into individual rows, so each hobby corresponds to the respective person's name. Ensure the resulting DataFrame has two columns: `Name` and `Hobby`.

from pyspark.sql import SparkSession
from pyspark.sql.functions import col, explode, split

# Initialize a Spark session
spark = SparkSession.builder.appName("HobbySeparator").getOrCreate()

# Sample data
data = [
    ("Ginny", "Reading, Cooking"),
    ("Mike", "Swimming, Reading"),
    ("Kevin", "Cooking, Travelling"),
    ("Amy", "Music, Painting")
]

# Define the schema
schema = ["Name", "Hobbies"]

# Create a DataFrame
df = spark.createDataFrame(data, schema)

# Split the 'Hobbies' column into an array and then explode it into separate rows
result_df = df.select(
    col("Name"),
    explode(split(col("Hobbies"), ", ")).alias("Hobby")
)

# Show the resulting DataFrame
result_df.show()
```
