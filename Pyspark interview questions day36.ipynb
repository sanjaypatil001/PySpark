```
#Find the number of students getting the same marks in both subjects (Maths and Science)

from pyspark.sql import SparkSession
from pyspark.sql.functions import col, count

# Initialize Spark session
spark = SparkSession.builder.appName("SameMarksCount").getOrCreate()

# Sample data
data = [
    (1, 85, 85),
    (2, 75, 70),
    (3, 90, 90),
    (4, 80, 85),
    (5, 85, 85),
    (6, 95, 90),
]
columns = ["student_id", "maths", "science"]
df = spark.createDataFrame(data, columns)

# Filter students with the same marks in maths and science
filtered_df = df.filter(col("maths") == col("science"))

# Count the number of students with the same marks
result = filtered_df.groupBy("maths", "science").agg(count("*").alias("number of students with same marks"))

# Show the result
result.show()
```
