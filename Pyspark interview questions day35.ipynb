```
#Write a PySpark program to find how many times a specific word appears in a text column of a DataFrame. For example, given the input text "how many times does the word how appear in this how string how", find how many times the word "how" appears.

from pyspark.sql import SparkSession
from pyspark.sql.functions import col, size, split

# Initialize Spark session
spark = SparkSession.builder.appName("WordCount").getOrCreate()

# Create a DataFrame
data = [("how many times does the word how appear in this how string how",)]
df = spark.createDataFrame(data, ["text"])

# Word to search
word = "how"

# Count occurrences of the word
df_with_count = df.withColumn(
    "word_count",
    size(split(col("text"), fr"\b{word}\b")) - 1
)

# Show the result
df_with_count.show()
```
