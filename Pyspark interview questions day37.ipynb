```
#Write a pyspark code to get the account balance for each account holder 

from pyspark.sql import SparkSession
from pyspark.sql.functions import col, sum, when

# Initialize Spark session
spark = SparkSession.builder.appName("AccountBalance").getOrCreate()

# Sample data
data = [
    (1, "credit", 500),
    (1, "debit", 200),
    (2, "credit", 1000),
    (2, "debit", 400),
    (1, "credit", 300)
]
columns = ["account_id", "transaction_type", "amount"]

# Create DataFrame
df = spark.createDataFrame(data, columns)

# Calculate account balance
account_balance = df.groupBy("account_id").agg(
    sum(when(col("transaction_type") == "credit", col("amount"))
        .when(col("transaction_type") == "debit", -col("amount"))
        .otherwise(0)).alias("account_balance")
)

# Show the result
account_balance.show()
```
