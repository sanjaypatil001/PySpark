```
#Write a PySpark code to calculate the number of new clients added to a company in each year based on their joining date?

from pyspark.sql import SparkSession
from pyspark.sql.functions import year, count

# Initialize Spark session
spark = SparkSession.builder.appName("NewClientsPerYear").getOrCreate()

# Sample data
data = [
    (1, "2020-03-15"),
    (2, "2021-07-20"),
    (3, "2021-11-12"),
    (4, "2022-01-05"),
    (5, "2020-08-19"),
    (6, "2022-04-25"),
    (7, "2023-09-13"),
]
columns = ["client_id", "joining_date"]

# Create DataFrame
df = spark.createDataFrame(data, columns)

# Extract year from joining_date and calculate count of new clients per year
new_clients_per_year = df.withColumn("year", year("joining_date")) \
    .groupBy("year") \
    .agg(count("client_id").alias("new_clients_count"))

# Show the result
new_clients_per_year.show()
```
